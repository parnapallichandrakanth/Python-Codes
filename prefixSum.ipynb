```python
A=[1,2,3,4,5]
currentProduct=1
prefixProduct=[]
for i in A:
  currentProduct*=i
  prefixProduct.append(currentProduct)
print(prefixProduct)
```

```python
A=[1,2,3,4,5]
CurrentValue=0
prefix_XOR_array=[]
for i in A:
  CurrentValue ^= i
  prefix_XOR_array.append(CurrentValue)
print(prefix_XOR_array)
```

#itertools.accumulate
- perform accumulations such as sums,products,or any other binary
- operation on a given array A (Prefix Array)


```python
import itertools
A=[10,20,30,40]
prefix_sum=list(itertools.accumulate(A))
print(prefix_sum)
```

```python
import itertools
def fun(a,b):
  return a*b
A=[10,20,30,40]
prefix_product=list(itertools.accumulate(A,func=fun))
print(prefix_product)
```

```python
import itertools
A=[1,2,3,4,5]
prefix_xor=list(itertools.accumulate(A,func=lambda a,b:a^b))
print(prefix_xor)
```

#Range Sum Queries
- We need to read the value of Q and given array(A)
- We need to print the sum of all elements from L to R  in A

```python
Q=int(input())
A=[10,20,30,40,50]
for i in range(Q):
  L,R=map(int,input().split())
  sm=0
  for j in range(L,R+1):
    sm+=A[j]
  #sm=sum(A[L:R+1])
  print(sm)

```

```python
#prefixSum range Query
A=[10,20,30,40,50]
prefix_s=[]
current_sum=0
Q=int(input())
for i in range(len(A)):
  current_sum+=A[i]
  prefix_s.append(current_sum)
for i in range(Q):
  L,R=map(int,input().split())
  if L==0:
    print(prefix_s[R])
  else:
    print(prefix_s[R]-prefix_s[L-1])


```

```python
#prefixSum range Query
A=[10,20,30,40,50]
prefix_s=[0]
#when we set first element as zero then no need to write if statement
#simply Sum(A[L...R])=prefix[R+1]-prefix[L]
#prefix_s=list(itertools.accumulate(A,initial=0))
current_sum=0
Q=int(input())
for i in range(len(A)):
  current_sum+=A[i]
  prefix_s.append(current_sum)
for i in range(Q):
  L,R=map(int,input().split())
  print(prefix_s[R+1]-prefix_s[L])
```
