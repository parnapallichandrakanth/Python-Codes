```python
a=[10,20,30,40]
for i in range(len(a)):
  for j in range(i,len(a)):
    print(i,j)
```

```python
a=[10,20,30,40]
N=len(a)
for start in range(N):
  for end in range(start,N):
    #above to loops responsible for genarate indexes like 0 to 0 and 0 to 1....etc
    # running a loop on start to end index and print subarray
    for i in range(start,end+1):
      print(a[i],end=" ")
    print()
```

```python

a="Chandu"
N=len(a)
sub_string_cnt=0
for start in range(N):
  for end in range(start,N):
    sub_string_cnt+=1
    for i in range(start,end+1):
      print(a[i],end=" ")
    print()
print(f"Sub-string count: {sub_string_cnt}")
```
