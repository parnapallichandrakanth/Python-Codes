                Creating a dictionary key value pairs




```python
d={}
n=int(input())
for i in range(n):
  key=input()
  value=int(input())
  d[key]=value
print(d)
```

                  Finding Which value is maximum


```python
d={'alice':18,'bob':27,'charlie':25}
mx=d['alice']
eldest=""
for key in d:
  if d[key]>mx:
    mx=d[key]
    eldest=key
print(mx)
print(eldest)

```

```python
d={'alice':18,'bob':27,'charlie':25}
mx=max(d.values())
print(mx)
for key in d:
  if d[key]==mx:
    print(key)
```

```python
s='chandrakanth'    #same code is applied to list of numbers and like that
d={}
for i in s:
  if i in d.keys():
    d[i]+=1
  else:
    d[i]=1
print(d)
```

#Accessing Keys and Values

```python
d={'chandu':19,'sarath':23}
print(d.keys())
print(d.values())
print(d.items())
for i in d.values():
  print(i)
for i in d.keys():
  print(i)
for i in d.items():
  print(i)
for i,j in d.items():
  print(i,j)
```
