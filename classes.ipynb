## constuctures
- in python a contructer is a special method
- that is used to instantiate the instance variables with values
- in python constucter is a dunder method
- it is a method that starts with __init__()
- contructer initialize object values when object is created

```python
class Student:
  def __init__(self,n,a,c):
    self.name=n
    self.cgpa=c
    self.age=a
s1=Student('Chandu',19,9.8)
print(s1.age)     #age of s1 (dot operation) -> obj_name.field_name
print(s1.cgpa)
print(s1.name)
```

```python
class Student:
  def __init__(self,n,a,c):
    self.name=n
    self.cgpa=c
    self.age=a
  def printDetails(self):
    print(f"Name is: {self.name}")
    print(f"Age is: {self.age}")
    print(f"CGPA is: {self.cgpa}")

s1=Student('Chandu',19,9.8)
s1.printDetails()
s2=Student("Sarath",23,9.7)
s2.printDetails()
```

```python
class Employee:
  def __init__(self,n,id,sa,de,a):
    self.name=n
    self.id=id
    self.salary=sa
    self.department=de
    self.age=a
  def printEmpDetails(self):
    print(f"Name of empolyee {self.name}")
    print(f"Employee id: {self.id}")
    print(f"Salary: {self.salary}")
    print(f"department: {self.department}")
    print(f"age: {self.age}")
e1=Employee('chandu',715,85000,'Automation',23)
e1.printEmpDetails()
e2=Employee('sarath',100,100000,'M achine Learning',24)
e2.printEmpDetails()
```

```python
class Rectangle:
  def __init__(self,l,w):
    self.length=l
    self.width=w
  def getArea(self):
    return self.length*self.width
  def getPerimeter(self):
    return 2*(self.length+self.width)
r1=Rectangle(20,10)
print(f"Area of s1 {r1.getArea()}")
print(f"perimeter of s1 {r1.getPerimeter()}")
r2=Rectangle(3,4)
print(f"Area of r2 is: {r2.getArea()}")
print(f"Perimeter of r2 is: {r2.getPerimeter()}")

```
