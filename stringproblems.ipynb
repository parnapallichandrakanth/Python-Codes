```python
s=input()
d={}
for i in s:
  d[i]=d.get(i,0)+1
min_=min(d.values())
max_=max(d.values())
a=abs(min_-max_)
c=0
for i in range(2,a):
  if a%i==0:
    c+=1
print("Diff: ",a)
if c==0:
  print("Absolute diff is Prime")
else:
  print("Abs diff is not prime num")



```
